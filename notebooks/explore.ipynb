```python
from pathlib import Path
import pandas as pd
import xarray as xr

PROJECT_ROOT = Path("/Users/brodiehasein/alberta_power_markets_project")

PREPROCESS_DIR = PROJECT_ROOT / "data" / "preprocessed"
WEATHER_DIR = PREPROCESS_DIR / "weather"

datasets = {
    "P&A_Table_preprocessed": PREPROCESS_DIR / "P&A_Table_preprocessed.csv",
    "interties_hour_ahead": PREPROCESS_DIR / "interties_hour_ahead.csv",
    "intertie_capability_preprocessed": PREPROCESS_DIR / "intertie_capability_preprocessed.csv",
    "generation_by_fuel_preprocessed": PREPROCESS_DIR / "generation_by_fuel_preprocessed.csv",
    "outages_preprocessed": PREPROCESS_DIR / "outages_preprocessed.csv",
    "wind_projects_preprocessed": PREPROCESS_DIR / "wind_projects_preprocessed.csv",
    "solar_projects_preprocessed": PREPROCESS_DIR / "solar_projects_preprocessed.csv",
}

print("=" * 100)
print("PREPROCESSED DATASET COLUMN INVENTORY")
print("=" * 100)

for name, path in datasets.items():
    df = pd.read_csv(path, nrows=0)

    print(f"\n{name}")
    print("-" * len(name))

    for col in df.columns:
        print(col)

print("\n" + "=" * 100)
print("WEATHER DATASET")
print("=" * 100)

weather_files = sorted(WEATHER_DIR.rglob("*.nc"))

print("Weather .nc files found:", len(weather_files))

if len(weather_files) == 0:
    print("No preprocessed weather .nc files found yet.")
    print("Checked:", WEATHER_DIR)
else:
    weather_file = weather_files[0]
    ds = xr.open_dataset(weather_file)

    print(f"Using: {weather_file}")

    for var in ds.data_vars:
        print(var)

    ds.close()
```
